## 1. Project Objective

Synthetic Sight is an image-classification project designed to distinguish between photographs of real human faces and AI-generated human-face images.

Our project will involve training, evaluating, and demonstrating a deep-learning model that predicts whether an uploaded face image is:

*   **Real:** A photograph of a real person
*   **Fake:** An AI-generated or synthetic face image

This project is intended as an educational machine-learning demonstration. It is not designed for production identity verification, criminal investigations, employment screening, or automated moderation decisions.

---

## 2. Research Question

> **How accurately can a ResNet-50 image-classification model distinguish real human-face photographs from AI-generated human-face images in the selected dataset?**

Our project will also explore:

*   Which types of classification errors occur most often
*   Whether the model performs differently across visible skin-tone groups
*   What limitations affect the model’s ability to generalize to unfamiliar images

---

## 3. Dataset

### Chosen Dataset

For this project, we will use the **140K Real and Fake Faces** dataset available through Kaggle.

The full dataset contains:

*   70,000 real face images derived from the Flickr-Faces-HQ dataset
*   70,000 AI-generated face images created using StyleGAN

## 4. Label Convention

We will use the following label convention throughout the project:

| Class | Numeric label | Meaning                              |
| :---- | ------------: | :----------------------------------- |
| Real  |           `0` | Photograph of a real human face      |
| Fake  |           `1` | AI-generated or synthetic human face |

**Fake is considered the positive class.**

All metrics, confusion matrices, prediction functions, charts, and presentation materials will consistently use this convention.

---

## 5. Model Choice

### Selected Model

Our chosen model for this project is:

> **ResNet-50 with transfer learning**

The model will leverage pretrained ImageNet weights and will be adapted for binary classification.

### Model Requirements

Our training workflow will include:

*   ResNet-50 architecture
*   Binary classification output
*   Training and validation tracking by epoch
*   Reproducible random seeding
*   Saving the best model checkpoint
*   Validation-based model selection
*   Final evaluation on the untouched test set

### Model Selection Strategy

The model checkpoint that shows the strongest validation performance will be selected as our final model.

The test set will not be used to guide decisions on:

*   The best training epoch
*   The classification threshold
*   Hyperparameters
*   Model architecture
*   Data augmentation settings

---

## 6. Classification Threshold

The default classification threshold will be:

```text
0.50
```

A predicted probability of `0.50` or greater will be classified as **Fake**.

A different threshold may be used only if:

1.  It is selected using validation-set results.
2.  The reason for changing it is documented.
3.  The test set is not used to select the threshold.
4.  The final threshold is applied consistently during test evaluation and demonstration.

## 7. Evaluation Metrics

### Primary Metric

Our primary evaluation metric will be:

> **F1-score for the Fake class**

The F1-score is our primary metric because it balances:

*   **Precision:** How often an image predicted as Fake is actually Fake
*   **Recall:** How many Fake images the model successfully identifies

## 10. Responsible Use Statement

Synthetic Sight is a research and educational prototype.

Its predictions should not be treated as definitive proof that an image is real or AI-generated. The model is trained on a limited dataset and may learn characteristics specific to FFHQ photographs, StyleGAN images, image compression, cropping, resolution, or other dataset artifacts.

The model may perform poorly on:

*   Images created by unfamiliar AI generators
*   Screenshots
*   Compressed social-media images
*   Edited photographs
*   Poor-quality images
*   Images without a clearly visible face
*   Multiple-person images
*   Images with unusual lighting or framing
*   Demographic groups that are underrepresented in the dataset

A prediction should always be presented as a model estimate rather than a verified fact.

## 11. Evaluation Approach

Our team will follow this sequence for evaluation:

1.  Finalize preprocessing and dataset splitting.
2.  Train the ResNet-50 model using the training set.
3.  Evaluate each epoch using the validation set.
4.  Save the best-performing validation checkpoint.
5.  Select the classification threshold using validation data.
6.  Freeze the model and threshold.
7.  Evaluate the frozen model once on the test set.
8.  Record all final metrics and visualizations.
9.  Conduct the responsible AI and error analysis.
10. Use the same frozen model in the final demonstration.

After final test evaluation begins, we will avoid repeatedly modifying and retesting the model based on test-set results.
